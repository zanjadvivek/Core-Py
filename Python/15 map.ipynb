The map() function is used to apply a function to each item in an iterable (like a list, tuple, etc.).

Use map() when you want to apply the same operation to every item in a list (or iterable).

####Syntax:
map(function, iterable)
- function → A function to apply (can be built-in, user-defined, or a lambda).

- iterable → A sequence (like a list, tuple, etc.)
 
- Returns a map object (which is an iterator) — usually converted to a list for visibility.

```python
def doublenum(x):
  return x*2
```

```python
lst = [1,2,3,4,5]
print(lst)
print(list(map(doublenum,lst)))
```

```python
countries = ["india","uk","us"]
print(countries)
print(list(map(lambda x: x.upper(),countries)))
```

```python
def converttoupper(y):
  return y.upper()
print(list(map(converttoupper,countries)))
```

####Using a Built-in Function

```python
numbers = ["1", "2", "3"]
result = map(int, numbers) 
print(list(result))         
```
