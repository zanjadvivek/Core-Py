### 1. String Indexing
Definition:
You can access characters in a string using index positions (starting from 0).

```python
text = "Python"
print(text[0])     # P
print(text[2])     # t
print(text[-1])    # n (last character)
```

####Looping Through a String

```python
name = "Vivek"
for char in name:
    print(char)
```

```python
i = 0
while i < len(name):
    print(name[i])
    i += 1
```

####find() Method
Purpose: Returns the index of the first occurrence of a substring.
If not found, returns -1 (safe to use).

```python
text = "hello world"
print(text.find("o"))  
print(text.find("z"))
```

####index() Method
Purpose: Same as find(), but raises an error if the substring is not found.

```python
text = "hello world"
print(text.index("o"))   
print(text.index("z")) 
```

####Escape Characters?
Definition:
Escape characters in strings are special sequences starting with a backslash (\) that allow you to represent characters that are not printable or would otherwise confuse the syntax (like quotes, tabs, newlines, etc.).

They are used to:

Include special characters in strings (like ", ')

Format text (like \n, \t)

Represent characters you cannot type directly

| Escape Code | Meaning                | Example               | Output               |
| ----------- | ---------------------- | --------------------- | -------------------- |
| `\'`        | Single quote (`'`)     | `'It\'s good'`        | It's good            |
| `\"`        | Double quote (`"`)     | `"She said, \"Hi!\""` | She said, "Hi!"      |
| `\\`        | Backslash (`\`)        | `"C:\\Users\\Vivek"`  | C:\Users\Vivek       |
| `\n`        | New line               | `"Line1\nLine2"`      | Line1 <br> Line2     |
| `\t`        | Tab (horizontal space) | `"Name:\tVivek"`      | Name:  Vivek         |
| `\r`        | Carriage return        | `"123\r456"`          | 456 (overwrites 123) |


###count() – Count Occurrences of Substring
Definition:
The count() method returns the number of times a substring appears in a string.

Syntax:

python
Copy
Edit
string.count(substring)

```python
text = "banana"
print(text.count("a"))   
print(text.count("an"))  
print(text.count("z"))   
```

```python
text = "banana"
print(text.count("a", 2, 5))  # Output: 1 (search from index 2 to 4)
```

####Strings Are Immutable
In Python, strings are immutable, meaning they cannot be changed after creation.

```python
s = "hello"
#s[0] = "H"   # ❌ Error: 'str' object does not support item assignment
```

```python
s = "hello"
s = "H" + s[1:]
print(s)
```
