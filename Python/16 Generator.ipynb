A generator is a function that yields values one at a time using the yield keyword instead of returning all at once.

It’s used to produce sequences lazily, which saves memory and improves performance especially for large datasets.

| Feature              | Benefit                                     |
| -------------------- | ------------------------------------------- |
| Memory Efficient     | Doesn’t store the entire sequence in memory |
| Lazy Evaluation      | Generates one item at a time as needed      |
| Infinite Sequences   | Can model endless data streams (like logs)  |
| Improves Performance | Especially with large data processing       |


```python
def custome_range(start,end):
  while start < end:
    yield start
    start += 1
for i in custome_range(1,10):
  print(i)
```

```python
def counter(n):
    i = 1
    while i<=n:
        yield i
        i += 1

getvalues = counter(5)
print(next(getvalues))
print(next(getvalues))
print(next(getvalues))
```

```python
j = (x*2 for x in range(10))

for i in j:
  print(i)
```
