A decorator is a function that modifies the behavior of another function without changing its code.

It allows you to wrap another function to add extra functionality before or after the original function runs.



```python
def my_decorator(func):
    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

```python
def convertto_upper(func):
    def inner():
        result = func()
        return result.upper()
    return inner

@convertto_upper
def greet():
    return "hello this is decoratore function"

print(greet())
```

```python
def square_num(func):
    def inner(a):
        i = func(a)
        return i * i
    return inner

@square_num
def num(x):
    return x

print(num(5))
```

```python
def square_allnum(func):
    def inner(a):
        result = func(a)
        return [i*i for i in result]
    return inner

@square_allnum
def num(x):
    return x

print(num([1,2,3,4]))
```

```python
def cube_num(func):
    def inner(a):
        i = func(a)
        return i**3
    return inner

@cube_num
def num(x):
    return x

print(num(3))
```

####Abstraction
Hiding the implementation details of a class and only showing the essential features to the user.

####Encapsulation
Wrapping data and functions into a single unit (object).
