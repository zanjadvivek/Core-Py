####Static method
Methods that don’t use the self parameter (work at class level)
- *Decorators allow us to wrap another function in order to
extend the behaviour of the wrapped function, without
permanently modifying it

```python
class Student:
    def __init__(self, name, rank):
        self.name = name
        self.rank = rank

    @staticmethod
    def greet():
        print("Congratulations")

    def getStudent(self):
        return f"{self.name}, your rank is {self.rank}"

s1 = Student("Vivek Zanjad", 1)
s1.greet()
print(s1.getStudent())
```

#####Create Account class with 2 attributes - balance & account no.Create methods for debit, credit & printing the balance

```python
class Account:
    def __init__(self, bal, acc):
        self.balance = bal
        self.account_no = acc

    #debit method
    def debit(self, amount):
        self.balance -= amount
        print(f"Amount {amount} debited from account {self.account_no}")
        print("total balance: ", self.balance)

    #credit method
    def credit(self, amount):
        self.balance += amount
        print(f"Amount {amount} credited to account {self.account_no}")
        print("total balance: ", self.balance)

    def get_balance(self):
        return self.balance

acc1 = Account(10000, 123456789)
acc1.debit(500)
acc1.credit(500)
acc1.credit(40000)
```

####Inheritance

When one class(child/derived) derives the properties & methods of another class(parent/base)

```python
#Inheritance example
class Car:
    color = "black"

    @staticmethod
    def start():
        print("car start")
     
    @staticmethod
    def stop():
        print("car stop")

class ToyotaCar(Car):
    def __init__(self, name):
        self.name = name

toyota = ToyotaCar("Toyota Fortuner")
print(toyota.name)
toyota.start()
toyota.stop()
```

```python
class Animal:
    def __init__(self, name):
        self.name = name
        print(f"Animal named {self.name} created")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # Calls Animal's __init__
        self.breed = breed
        print(f"{self.name} is a {self.breed}")
    
d = Dog("Buddy", "Golden Retriever")
```

####Multi-level Inheritance
Multilevel Inheritance refers to a scenario where a class is derived from a child class, which in turn is derived from another parent class.

```python
class Car:
    @staticmethod
    def start():
        print("car start")
     
    @staticmethod
    def stop():
        print("car stop")

class ToyotaCar(Car):      
    def __init__(self, brand):
        self.brand = brand

class Fortuner(ToyotaCar):
    def __init__(self, brand, type):
        super().__init__(brand)
        self.type = type

car1 = Fortuner("Fortuner","Disel")
print(car1.brand)
print(car1.type)
car1.start()
car1.stop()
```

####Multiple Inheritance
Multiple inheritance is when a class inherits from more than one parent class.

This allows the derived class to access the attributes and methods of all parent classes.

```python
class Father:
    def skills(self):
        print("Father: Gardening, Programming")

class Mother:
    def skills(self):
        print("Mother: Cooking, Art")

class Child(Father, Mother):  # Inherits from both
    def skills(self):
        print("Child:", end=" ")
        Father.skills(self)
        Mother.skills(self)

# Create object
c = Child()
c.skills()
```

####super() Method
The super() function is used to call a method from the parent class — usually the constructor (__init__) — in a way that works well with inheritance.

It is commonly used to initialize the parent class when creating child classes.

super() method is used to access methods of the parent class

```python
class Parent:
    def __init__(self):
        print("Parent Constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()  # Calls Parent's __init__()
        print("Child Constructor")

c = Child()
```

####Class Method
A class method is bound to the class and receives the class as an implicit first argument

Note- static method can't access or modify class state and generally for utility

```python
class Person:
    name = "xyz"

    @classmethod
    def get_name(cls, name):
        cls.name = name

p1 = Person()
print(p1.name)
p1.get_name("abc")
print(p1.name)
print(Person.name)
```

####Property decoratore
We use @property decoratores on any method in the class to use the method as  a property

```python
class Student:
    def __init__(self, phy, chem, maths):
        self.phy = phy
        self.chem = chem
        self.maths = maths

    @property
    def percentage(self):
        return str((self.phy + self.chem + self.maths)/3) + "%"
    
stu1 = Student(70, 80, 90)
print(stu1.percentage)

stu1.chem = 86
print(stu1.percentage)
```

When the same operator is allowed to have diffrent meaning according to the text

####Built-in Operators and Their Dunder Methods
| Operator | Description           | Dunder Method              | Example                    |
| -------- | --------------------- | -------------------------- | -------------------------- |
| `+`      | Addition              | `__add__(self, other)`     | `x + y → x.__add__(y)`     |
| `-`      | Subtraction           | `__sub__(self, other)`     | `x - y → x.__sub__(y)`     |
| `*`      | Multiplication        | `__mul__(self, other)`     | `x * y → x.__mul__(y)`     |
| `/`      | Division              | `__truediv__(self, other)` | `x / y → x.__truediv__(y)` |
| `%`      | Modulus (remainder)   | `__mod__(self, other)`     | `x % y → x.__mod__(y)`     |
| `==`     | Equality check        | `__eq__(self, other)`      | `x == y → x.__eq__(y)`     |
| `<`      | Less than             | `__lt__(self, other)`      | `x < y → x.__lt__(y)`      |
| `len()`  | Length of object      | `__len__(self)`            | `len(x) → x.__len__()`     |
| `str()`  | String representation | `__str__(self)`            | `str(x) → x.__str__()`     |


```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img
    
    def showNumber(self):
        print(self.real,"i +" , self.img, "j")

    def __add__(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)

num1 = Complex(1,3)
num1.showNumber() # 1 + 3j

num2 = Complex(2,4)
num2.showNumber()

num3 = num1 + num2
num3.showNumber()
```

```python
class Circle:
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return (22/7)*self.radius ** 2

    def perimeter(self):
        return 2*(22/7)*self.radius    
    
c1 = Circle(21)
print(c1.area())
print(c1.perimeter())
```
