A function is a reusable block of code that performs a specific task.
It helps make code modular, reusable, and clean.

```python
def greet():
    print("Hello, welcome to Python!")

greet()  # Output: Hello, welcome to Python!
```

####Function with Parameters

```python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)  # Output: 8
```

####Default Parameters

```python
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet("Vivek")     # Output: Hello, Vivek!
greet()            # Output: Hello, Guest!
```

####Return Multiple Values


```python
def calculate(a, b):
    return a + b, a - b

sum_, diff = calculate(10, 5)
print(sum_, diff)  # Output: 15 5
```

####Local Scope

```python
def my_func():
    x = 10  # Local variable
    print(x)

my_func()        # Output: 10
# print(x)       # ❌ Error: x is not defined outside
```

####Global Scope

```python
x = 100  # Global variable

def show():
    print(x)  # Can access global variable

show()       # Output: 100
print(x)     # Output: 100
```

####Modifying Global Variables (using global)

```python
def msg():
    global a
    a = 10
    print(a)
msg()
print(a)
```

```python
count = 5

def increment():
    global count
    count += 1

increment()
print(count)  # Output: 1
```

#### Enclosing Scope (Nested Functions)

```python
def outer():
    x = "outer"
    
    def inner():
        print(x)  # Accesses enclosing variable
    
    inner()

outer()  # Output: outer
```

####Arguments
In Python, you can pass values to functions using arguments (or parameters)

####1. Positional Arguments

```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old.")

greet("Alice", 25)
# Output: Hello Alice, you are 25 years old.
```

####2. Keyword Arguments
You can specify argument names to make the call clearer and ignore the order.

```python
greet(age=30, name="Bob")
# Output: Hello Bob, you are 30 years old.
```

####3. Default Arguments
You can give a default value to a parameter.

```python
def greet(name, age=18):
    print(f"Hi {name}, age {age}")

greet("John")         # age defaults to 18
greet("Sara", 22)     # overrides default
```

####4. Arbitrary Arguments (*args)
Used when you don’t know how many arguments will be passed.

It collects positional arguments into a tuple.

```python
def add_numbers(*args):
    print(args)
    return sum(args)

print(add_numbers(1, 2, 3, 4))  # Output: 10
```

```python
def EmpDetails(Empid,Empname,*args,city="mumbai"):
    print(Empid,Empname,sum(args),city)

EmpDetails(101,"Amit",10000,20000,city="Delhi")
```

####5.Arbitrary Keyword Arguments (**kwargs)
Used when you don’t know how many keyword arguments will be passed.

It collects them into a dictionary.

```python
def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Amit", age=22, city="Pune")
```

```python
def EmpDetails(**kwargs):
    print(kwargs.values())
EmpDetails (name="Amit", age=22, city="Pune")
```

```python
def EmpDetails(Empid,Empname,*args,city="mumbai",**kwargs):
    print(Empid,Empname,sum(args),city,kwargs)

EmpDetails(101,"Amit",10000,20000,city="Delhi",salary=100000,email="abc@gmail.com")
```

####6. Positional-only and Keyword-only Arguments (Advanced – optional)

```python
def example(a, /, b, *, c):
    print(a, b, c)

example(1, b=2, c=3)  # ✅
# example(a=1, b=2, c=3)  ❌ 'a' must be positional-only
```

####Python Docstring (Documentation String)
A docstring is a string literal used to document a function, class, module, or method.
It helps explain what the code does, how to use it, and what to expect.

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add.__doc__)
```
