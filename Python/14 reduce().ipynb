The reduce() function is used to apply a rolling computation to a sequence, reducing it to a single value.

📌 reduce() is part of the functools module.

Use reduce() when you want to combine all elements into one single value.

👉 Think: “Accumulate or aggregate.”

Examples:

Sum
Product
Finding max
Building one result from many

####Example 1: Sum of List Elements

```python
from functools import reduce

numbers = [1, 2, 3, 4]

result = reduce(lambda x, y: x + y, numbers)
print(result) 
```

####Example 2: Multiply All Elements

```python
from functools import reduce

nums = [2, 3, 4]
result = reduce(lambda x, y: x * y, nums)
print(result)  
```

```python
from functools import reduce

nums = [2, 3, 4]
result = reduce(lambda x,y: x + y, nums)
print(result)
```

####Find the Maximum Value in a List

```python
from functools import reduce

nums = [10,20,30,50,90]
max_value = reduce(lambda a,b: a if a > b else b, nums)
print(max_value)
```

####Concatenate a List of Strings

```python
from functools import reduce

words = ["Python", "is", "awesome"]
sentence = reduce(lambda x,y: x + " "+ y,words)
print(sentence)
```

####Compute Factorial of a Number

```python
from functools import reduce

n = 5  # 5! = 5 × 4 × 3 × 2 × 1
factorial = reduce(lambda x, y: x * y, range(1, n + 1))
print(factorial)  
```

####Flatten a List of List

```python
from functools import reduce

nested = [[1, 2], [3, 4], [5]]

flattened = reduce(lambda x, y: x + y, nested)
print(flattened) 
```

####Find Product of All Even Numbers

```python
from functools import reduce

nums = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x%2 == 0, nums))
product = reduce(lambda x,y: x*y, even)
print(product)
```

####Merge List of Dictionaries

```python
from functools import reduce

dicts = [{'a': 1}, {'b': 2}, {'c': 3}]

merged = reduce(lambda x, y: {**x, **y}, dicts)
print(merged)
```
