None is a special constant in Python that represents the absence of a value or a null value.



```python
x = None
print(x)          # Output: None
print(type(x))    # Output: <class 'NoneType'>
```

```python
def greet():
    print("Hello!")

result = greet()
print(result)     # Output: Hello! then None (because no return)
```

```python
x = None

if x is None:
    print("x is None")

if x == None:
    print("x equals None (not recommended, use 'is')")
```

```python
def process(data=None):
    if data is None:
        data = []
    data.append("item")
    return data

print(process())  # ['item']
```
