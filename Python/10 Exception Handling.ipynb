Exception handling allows you to gracefully handle errors in your code so that your program doesn't crash unexpectedly.

####Keywords Used
| Keyword   | Description                                |
| --------- | ------------------------------------------ |
| `try`     | Block of code to test for errors           |
| `except`  | Block to handle the error                  |
| `else`    | Block to run if no error occurs            |
| `finally` | Block that always runs (error or no error) |


```python
#Syntax
try:
    # Code that may cause error
except ExceptionType:
    # Code to handle the error
else:
    # Code to run if no error occurs
finally:
    # Always runs
```

```python
try:
    a = int(input("Enter a number: "))
    print(10 / a)
except ValueError:
    print("Please enter a valid number!")
except ZeroDivisionError:
    print("You can't divide by zero!")
else:
    print("Everything went fine!")
finally:
    print("Program ends.")
```

####Catching Multiple Exceptions

```python
try:
    num = int("abc")
    result = 10 / 0
except (ValueError, ZeroDivisionError) as e:
    print("An error occurred:", e)
```

####Using raise to Throw Exception

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age can't be negative")
    return age

set_age(-5)  # This raises ValueError
```
