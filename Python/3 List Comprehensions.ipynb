A list comprehension is a short and elegant way to create a new list by applying an expression to each item in an iterable (like a list, tuple, or range).

It’s faster, cleaner, and more Pythonic than using loops.

#####Basic Syntax
[expression for item in iterable if condition]

1.expression: The operation or value to include in the list.

2.item: The variable that takes the value from iterable.

3.iterable: Any iterable (like a list, range, string, etc.).

4.condition (optional): Filters items (like if item > 10).

####Example 1: Square Numbers from 1 to 5

```python
squares = [x**2 for x in range(1, 6)]
print(squares)  
```

####Example 2: Create a List of Even Numbers

```python
evens = [x for x in range(10) if x % 2 == 0]
print(evens)  
```

####Example 3: Convert to Uppercase

```python
fruits = ['apple', 'banana', 'mango']
upper_fruits = [fruit.upper() for fruit in fruits]
print(upper_fruits)
```

####Example 4: Use if-else inside List Comprehension

```python
nums = [1, 2, 3, 4, 5]
labels = ["even" if x % 2 == 0 else "odd" for x in nums]
print(labels)
```

####Example 5: Nested Loop in List Comprehension

```python
pairs = [(x, y) for x in [1, 2] for y in [3, 4]]
print(pairs) 
```
