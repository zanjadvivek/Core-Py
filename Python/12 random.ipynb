The random module in Python is used to generate random numbers, choose random elements, shuffle data, etc.



```python
#Import
import random
```

####Common Functions in random
| Function                   | Description                                               | Example                         |
| -------------------------- | --------------------------------------------------------- | ------------------------------- |
| `random.random()`          | Returns a random float from 0.0 to < 1.0                  | `random.random()` → `0.6731...` |
| `random.randint(a, b)`     | Returns a random **integer** between a and b (inclusive)  | `random.randint(1, 6)` → `3`    |
| `random.uniform(a, b)`     | Returns a random **float** between a and b                | `random.uniform(1.5, 5.5)`      |
| `random.choice(seq)`       | Randomly chooses an item from a list/tuple/string         | `random.choice(['a','b','c'])`  |
| `random.choices(seq, k=n)` | Returns a list of **k** random choices (with replacement) | `random.choices([1,2,3], k=2)`  |
| `random.sample(seq, k=n)`  | Returns a list of **k** unique elements (no duplicates)   | `random.sample(range(10), 3)`   |
| `random.shuffle(seq)`      | Shuffles a **list** in place                              | `random.shuffle(my_list)`       |


####Generate Random Float Between 0 and 1

```python
print(random.random())  # Output: 0.2756...
```

####Generate Random Integer Between 10 and 20

```python
print(random.randint(10, 20))  
```

####Choose Random Item From a List

```python
fruits = ["apple", "banana", "cherry"]
print(random.choice(fruits)) 
```

####Shuffle a List

```python
nums = [1, 2, 3, 4, 5]
random.shuffle(nums)
print(nums)  
```

#### Random Sample (No Repeats)

```python
print(random.sample(range(100), 5))  
```
