Recursion is a programming technique where a function calls itself to solve smaller instances of the same problem.

```python
def printnums(n):
  if n > 0:
    printnums(n-1)
    print(n)

printnums(10)
```

```python
def sumNumbers(n):
    total = 0
    for i in range(1,n+1):
        total += i
    return total
sumNumbers(10)
```

```python
def sumNumbersRec(x):
    if x == 1: 
        return 1
    else:
        return x + sumNumbersRec(x-1)
    
sumNumbersRec(10)
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)
factorial(5)
```

####Fibonacci Series

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
print(fibonacci(10))
```

####Reverse a String Recursively

```python
def reverse(s):
    if len(s) == 0:
        return ""
    else:
        return s[-1] + reverse(s[:-1])

print(reverse("hello"))  
```

####Sum of Digits 
Q: Write a function that returns the sum of digits of a number using recursion.
Input: 1234 → Output: 10

```python
def sum_of_digits(n):
    if n == 0:
        return 0
    else:
        return n % 10 + sum_of_digits(n // 10)
print(sum_of_digits(12345))
```

####Reverse string

```python
def reverser_string(s):
    if len(s) == 0:
        return ""
    else:
        return s[-1] + reverser_string(s[:-1])
print(reverser_string("hello"))
```

####Palindrome string

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    else:
        return s[0] == s[-1] 
    return is_palindrome(s[1:-1])
print(is_palindrome("bob"))
```

####Palindrome number

```python
def palindrome_num(n):
    original = n
    reversed_num = 0
    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n = n // 10
    return original == reversed_num
print(palindrome_num(121))
```

```python
def reverse_number(n, rev=0):
    if n == 0:
        return rev
    digit = n % 10
    return reverse_number(n // 10, rev * 10 + digit)
print(reverse_number(121))
```

#### Count Ways to Climb Stairs (Medium)
Q: Given n stairs, return number of ways to reach the top taking 1 or 2 steps at a time.


```python
def climb_stairs(n):
    if n <= 1:
        return 1
    return climb_stairs(n-1)  + climb_stairs(n-2)
print(climb_stairs(4))
```

####Prime number

```python
def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

# Example usage
num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```
