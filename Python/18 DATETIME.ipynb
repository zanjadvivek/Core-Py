The datetime module in Python allows you to work with dates and times — including getting the current date/time, formatting, arithmetic, and parsing.

###DATETIME MODULE

The datetime module is a built-in Python module that provides classes and functions to manipulate dates and times.

```python
#must import it:
import datetime
```

| Class       | Description                                 |
| ----------- | ------------------------------------------- |
| `date`      | Represents **date** (year, month, day)      |
| `time`      | Represents **time** (hour, min, sec, micro) |
| `datetime`  | Combines `date` + `time`                    |
| `timedelta` | Represents a **time duration/difference**   |


####datetime.date – Working with Dates


```python
from datetime import date

today = date.today()
print(today)        # e.g., 2025-07-07
print(today.year)   # 2025
print(today.month)  # 7
print(today.day)    # 7
```

#### Create a custom date

```python
d = date(2025, 12, 25)
print(d)  
```

####datetime.time – Working with Time

```python
from datetime import time

t = time(14, 30, 15)  
print(t)              
```

####datetime.datetime – Full Date + Time

```python
from datetime import datetime

now = datetime.now()
print(now)           
```

####Create a specific datetime

```python
dt = datetime(2025, 12, 25, 10, 30)
print(dt)  
```

####datetime.timedelta – Date Arithmetic

```python
from datetime import datetime, timedelta

now = datetime.now()
next_week = now + timedelta(days=7)
yesterday = now - timedelta(days=1)

print(next_week)
print(yesterday)
```

####Formatting Dates – strftime()

strftime() = String Format Time

It converts a datetime or date object into a formatted string.

```python
from datetime import datetime

now = datetime.now()
formatted = now.strftime("%d-%m-%Y %H:%M:%S")
print(formatted)  
```

####Parsing Strings to Date – strptime()

strptime() = String Parse Time

It converts a string into a datetime object using a known format.

```python
from datetime import datetime

date_str = "25-12-2025"
dt_obj = datetime.strptime(date_str, "%d-%m-%Y")
print(dt_obj) 
```

####Get Day Name, Weekday, etc.

```python
today = datetime.today()

print(today.strftime("%A"))  
print(today.weekday())        
```

####Example Mini Use-Case

```python
from datetime import datetime, timedelta

birth = datetime(2003, 1, 15)
today = datetime.now()
age = today - birth

print("Days alive:", age.days)
```

####timedelta – Time Duration or Difference
A class from the datetime module.

Represents a duration — difference between two datetime objects.

Can be added to or subtracted from dates.

####Syntax
from datetime import timedelta

delta = timedelta(days=5, hours=2, minutes=30)


####Example 1: Add Days to a Date

```python
from datetime import datetime, timedelta

today = datetime.now()
future = today + timedelta(days=7)
print("Today:", today)
print("7 Days Later:", future)
```

####Example 2: Calculate Difference Between Two Dates

```python
from datetime import datetime

d1 = datetime(2025, 7, 1)
d2 = datetime(2025, 7, 7)

diff = d2 - d1
print("Difference:", diff)         # 6 days, 0:00:00
print("Days only:", diff.days)     # 6
```

####timestamp() – Convert Datetime to POSIX Timestamp
A Unix timestamp (also called POSIX time) is the number of seconds since Jan 1, 1970 UTC.

It's commonly used in databases, APIs, and time comparisons.

####datetime.timestamp()

```python
from datetime import datetime

dt = datetime(2025, 7, 7, 12, 0, 0)
ts = dt.timestamp()
print("Timestamp:", ts)
```

#### Convert timestamp back to datetime

```python
from datetime import datetime

ts = 1751899200.0
dt = datetime.fromtimestamp(ts)
print("Datetime:", dt)
```

####Example Use Case (Combined)

```python
from datetime import datetime, timedelta

login_time = datetime.now()
logout_time = login_time + timedelta(hours=3)

duration = logout_time - login_time
print("Session duration:", duration)

timestamp = login_time.timestamp()
print("Login as timestamp:", timestamp)
```

To work with months, you can use the relativedelta class from the dateutil module (not monthdelta directly, unless using a 3rd-party package).

#### Example: Add or Subtract Months

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

date = datetime(2025, 7, 16)

# Add 2 months
future = date + relativedelta(months=2)

# Subtract 3 months
past = date - relativedelta(months=3)

print("Original:", date)
print("+2 Months:", future)
print("-3 Months:", past)
```

####Add Years, Days, Weeks

```python
date = datetime(2025, 9, 26)

print(date + relativedelta(years=1))     # 2026-07-16
print(date + relativedelta(weeks=2))     # 2025-07-30
```

```python
from monthdelta import monthmod
import datetime

d1 = datetime.date(2021, 1, 12)
d2 = datetime.date(2023, 10, 22)

md, rd = monthmod(d1, d2)
print(md.months)    # Output: 33 (months between)
print(rd.days)      # Output: 10  (remaining days)
```

###Common Format Codes

| Format Code | Meaning              | Example Output |
| ----------- | -------------------- | -------------- |
| `%d`        | Day (01–31)          | 16             |
| `%m`        | Month (01–12)        | 07             |
| `%Y`        | Year (4-digit)       | 2025           |
| `%y`        | Year (2-digit)       | 25             |
| `%H`        | Hour (00–23)         | 14             |
| `%I`        | Hour (01–12)         | 02             |
| `%p`        | AM or PM             | PM             |
| `%M`        | Minute (00–59)       | 35             |
| `%S`        | Second (00–59)       | 00             |
| `%A`        | Weekday name         | Wednesday      |
| `%a`        | Weekday abbreviation | Wed            |
| `%B`        | Month name           | July           |
| `%b`        | Month abbreviation   | Jul            |
