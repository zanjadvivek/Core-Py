A lambda function in Python is a short, anonymous function that is defined without a name, used for simple operations.

Use a lambda function when you need a small, temporary function that you don’t want to formally define with def.

####Basic Syntax:
lambda arguments: expression

- lambda is the keyword

- You can have any number of arguments, but only one expression
 
- The result of the expression is automatically returned

####Example 1: Add Two Numbers

```python
add = lambda a, b: a + b
print(add(3, 4))   # Output: 7
```

```python
square = lambda x: x * x
print(square(5))   # Output: 25
```

#####Use in map()

```python
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x**2, nums))
print(squares)  # Output: [1, 4, 9, 16]
```

####Use in filter()

```python
nums = [1, 2, 3, 4, 5]
even = list(filter(lambda x: x % 2 == 0, nums))
print(even)  # Output: [2, 4]
```

####Use in sorted()

```python
students = [("Alice", 21), ("Bob", 19), ("Charlie", 23)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
```
