```python
#1. Right-Angled Triangle
n = 5
for i in range(1, n+1):
    print("* " * i)
```

```python
# 2. Inverted Right-Angled Triangle
n = 5
for i in range(n, 0, -1):
    print("* " * i)
```

```python
#3. Pyramid Pattern
n = 5
for i in range(1, n+1):
    print(" " * (n-i) + "* " * i)
```

```python
#4. Number Triangle
n = 5
for i in range(1, n+1):
    for j in range(1, i+1):
        print(j, end=" ")
    print()
```

```python
#5. Floyd’s Triangle
n = 4
num = 1
for i in range(1, n+1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
#6. Diamond Pattern
n = 5
# Upper part
for i in range(1, n+1):
    print(" " * (n-i) + "* " * i)
# Lower part
for i in range(n-1, 0, -1):
    print(" " * (n-i) + "* " * i)
```
